Тема: первичный разведочный анализ датасета *"FMCG Sales & Profit Dataset (2023–2025)"* для дальнейшей коммерческой аналитики

```python
# Импорт нужных библиотек
import pandas as pd
import matplotlib.pyplot as plt

# Стиль для графиков
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
# Загрузка данных
df = pd.read_csv('../data/raw-data/fmcg_sales_marketing_profitability_2023_2025.csv')
df.head() 
```

```python
# Получение общей информации о датафрейме
df.info()
```

```python
# Проверка на пропущенные значения
missing_values = df.isnull().sum()  # Подсчет пропущенных значений в каждом столбце
missing_values = missing_values[missing_values > 0].reset_index().rename(columns={'index': 'Column', 0: 'Missing Values'})  # Фильтрация и переименование столбцов
if missing_values.empty:
    print("\033[1m" + '- ПРОПУСКОВ В ДАННЫХ НЕ ОБНАРУЖЕНО!' + "\033[1m")
else:
    display(missing_values)
```

```python
# Проверка на дубликаты в данных
duplicate_count = df.duplicated().sum()
if duplicate_count == 0:
    print("\033[1m" + '- ДУБЛИКАТОВ В ДАННЫХ НЕ ОБНАРУЖЕНО!')
else:
    print(duplicate_count)
print()
```

```python
# Определение периода, охватываемого данными
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

start = df["Order_Date"].min()
end = df["Order_Date"].max()
years = (end - start).days / 365.25

print(
    f"Данные охватывают период с {start.date()} по {end.date()}. "
    f"Продолжительность — примерно {years:.1f} года."
)
```

```python
geo_summary = df.groupby([df.Region, df.Country]).agg({'City': 'nunique', 'Sales_Person': 'nunique'}).reset_index()
print(f'Данные охватывают {df.Region.nunique()} региона, {df.Country.nunique()} стран, {df.City.nunique()} городов и {df.Sales_Person.nunique()} торговых представителей.')
geo_summary
```

```python
print(f'Данные охватывают {df.Product_Category.nunique()} категорий продуктов, {df.Brand.nunique()} брендов продуктов и самих {df.Product_Name.nunique()} уникальных продуктов.')
```

```python
print(f'В данных представлено {df.Customer_Type.nunique()} типа клиентов, {df.Sales_Channel.nunique()} канала продаж и {df.Promotion_Type.nunique() - 1} различных типов промо-акций (Также значение "No Promo").')
```

```python
# Денежные показатели
gross_sales = df.Gross_Sales_USD.sum()
net_revenue = df.Net_Revenue_USD.sum()
units_sold = df.Units_Sold.sum()
discount_load = (1 - net_revenue / gross_sales) * 100 if gross_sales != 0 else 0
marketing_spend = df.Marketing_Spend_USD.sum()
logistics_cost = df.Logistics_Cost_USD.sum()
COGS = df.COGS_USD.sum()
profit = net_revenue - (marketing_spend + logistics_cost + COGS)
profit_margin = profit / net_revenue * 100 if net_revenue != 0 else 0

print(f'\nВаловая выручка ${gross_sales:_}'.replace('_', ' '),
      f'\nСкидочная нагрузка {discount_load:.2f}%, потрачено ${gross_sales - net_revenue:,.2f} на скидки'.replace(',', ' '),
      f'\nВыручка после скидок ${net_revenue:,.2f}'.replace(',', ' '),
      f'\nМаркетинговые расходы ${marketing_spend:,.2f} ({marketing_spend / net_revenue * 100:,.2f}% от выручки после скидок)'.replace(',', ' '),
      f'\nЛогистические расходы ${logistics_cost:,.2f} ({logistics_cost / net_revenue * 100:,.2f}% от выручки после скидок)'.replace(',', ' '),
      f'\nCOGS ${COGS:,.2f} ({COGS / net_revenue * 100:,.2f}% от выручки после скидок)'.replace(',', ' '),
      f'\nПрибыль ${profit:,.2f}, маржинальность {profit_margin:.2f}%'.replace(',', ' '),
      f'\nПродано единиц {units_sold:,}'.replace(',', ' ')
)
```
